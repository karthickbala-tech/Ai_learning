### **Call api using get methods**

```python
import requests
```

```python
def call_api(url):
    try:
        response=requests.get(url)
        response.raise_for_status()
        data=response.json()
        print("Response JSON:")
        if not data:
            print("Data not found")
        elif isinstance(data, list):
            for item in data:
                print(item)
        else:
            print(data)
    except requests.exceptions.RequestException as e:
        print("Error:", e)
```

```python
call_api("https://api.restful-api.dev/objects")
```

```python
call_api("https://api.restful-api.dev/objects/7")
```

```python
call_api("https://api.restful-api.dev/objects?id=3&id=5&id=10")
```

### **Add data using post methods**

```python
url="https://api.restful-api.dev/objects"

payload = {
   "name": "Apple MacBook Pro 099",
   "data": {
      "year": 2023,
      "price": 2889.99,
      "CPU model": "Intel Core i9",
      "Hard disk size": "2 TB"
   }
}

headers={
    "Content-Type": "application/json"
}

response=requests.post(url, json=payload, headers=headers)

if response.status_code in [200, 201]:
    print("Data added successfully")
    print(response.json())
else:
    print("Failed to add data")
    print(response.status_code)

```

### **Put methods**

```python
url="https://api.restful-api.dev/objects"

payload = {
   "name": "Apple MacBook Pro 009",
   "data": {
      "year": 2023,
      "price": 2889.99,
      "CPU model": "Intel Core i9",
      "Hard disk size": "2 TB"
   }
}

headers={
    "Content-Type": "application/json"
}

response=requests.post(url, json=payload, headers=headers)

if response.status_code in [200, 201]:
    print("Data added successfully")
    print(response.json())
else:
    print("Failed to add data")
    print(response.status_code)


object_id = "ff8081819c5368bb019c56ad4f9b0775"  # example id
url = f"https://api.restful-api.dev/objects/{object_id}"

payload = {
   "name": "Apple MacBook Pro Updated",
   "data": {
      "year": 2023,
      "price": 2499.99,
      "CPU model": "M3 Pro",
      "Hard disk size": "2 TB"
   }
}

response = requests.put(url, json=payload)

if response.status_code == 200:
    print("Data updated successfully")
    print(response.json())
else:
    print("Failed to update")
    print(response.status_code)

```
