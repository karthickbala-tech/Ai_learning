```python
import pdfplumber
import re

def extract_abstract_and_conclusion(pdf_path):
    text = ""

    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"

    text = re.sub(r'\s+', '\n', text)

    abstract_pattern = re.compile(
        r'abstract(.*?)(introduction|background|methods)',
        re.IGNORECASE | re.DOTALL
    )

    conclusion_pattern = re.compile(
        r'(conclusion|conclusions)(.*?)(references|bibliography|acknowledgements|appendix|$)',
        re.IGNORECASE | re.DOTALL
    )

    abstract_match = abstract_pattern.search(text)
    conclusion_match = conclusion_pattern.search(text)

    abstract = abstract_match.group(1).strip() if abstract_match else "Abstract not found."
    conclusion = conclusion_match.group(2).strip() if conclusion_match else "Conclusion not found."

    return abstract, conclusion
```

```python
pdf_file = "/content/1810.04805v2.pdf"

abstract, conclusion = extract_abstract_and_conclusion(pdf_file)

print("\n========== ABSTRACT ==========\n")
print(abstract)

print("\n========== CONCLUSION ==========\n")
print(conclusion)
```

```python
def extract_text(pdf_path):
    doc = fitz.open(pdf_path)
    text = ""

    for page in doc:
        text += page.get_text()

    doc.close()
    return text

```

```python
def clean_text(text: str) -> str:

    text = re.sub(r'\bPage\s*\d+\b', '', text, flags=re.IGNORECASE)
    text = re.sub(r'\b\d+\b', '', text)


    headers_footers = ["Confidential", "Report", "Draft", "FooterText"]
    for hf in headers_footers:
        text = re.sub(rf'\b{hf}\b', '', text, flags=re.IGNORECASE)
    text = re.sub(r'\s+', '', text)
    text = re.sub(r'[^\w\s.,;:!?-]', '', text)
    return text.strip()
```

```python
import re

def fix_spacing(text: str) -> str:

    text = re.sub(r'([a-z])([A-Z])', r'\1 \2', text)
    text = re.sub(r'([a-zA-Z])(\d)', r'\1 \2', text)
    text = re.sub(r'(\d)([a-zA-Z])', r'\1 \2', text)

    return text
```

```python
def restore_structure(text: str) -> str:
    sections = [
        "Abstract", "Introduction", "Related Work", "Method",
        "Model Architecture", "Pre-training", "Fine-tuning",
        "Experiments", "Results", "Conclusion", "References"
    ]

    for sec in sections:
        text = re.sub(
            rf'\b{sec}\b',
            f'\n\n{sec}\n',
            text,
            flags=re.IGNORECASE
        )

    # sentence spacing
    text = re.sub(r'\.\s*', '.\n', text)

    return text.strip()
```

```python
def clean_text(text: str) -> str:
    text = re.sub(r'\bPage\s*\d+\b', '', text, flags=re.IGNORECASE)
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'[^\w\s.,;:!?()\-%/]', '', text)
    return text.strip()
```

```python
summarizer = pipeline(
    task="summarization",
    model="facebook/bart-large-cnn",
    device=-1  # CPU
)
```

```python
def summarize_text(text, max_len=200):
    max_chunk_size = 900
    chunks = []

    for i in range(0, len(text), max_chunk_size):
        chunks.append(text[i:i + max_chunk_size])

    summary = ""

    for chunk in chunks:
        result = summarizer(
            chunk,
            max_length=max_len,
            min_length=50,
            do_sample=False
        )
        summary += result[0]["generated_text"] + " "

    return summary.strip()
```
